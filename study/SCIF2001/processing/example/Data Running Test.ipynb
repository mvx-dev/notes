Using the readPTU library, using these instructions https://github.com/QuantumPhotonicsLab/readPTU/tree/master

```python
import readPTU #Import the library, read the file and print the header to check that it actually worked.
from matplotlib import pyplot as plt
ptu_file = readPTU.PTUfile(r'AtB_10s_20mW_29Oct1.ptu')
ptu_file.print_header()
```

```python
ptu_meas = readPTU.PTUmeasurement(ptu_file) #Reads in a file into a measurement class, so we can do measurements on the data 
```

```python
one, two, three = ptu_meas.timetrace(channel=1)
plt.plot(one,two)
one, two, three = ptu_meas.timetrace(channel=2)
plt.plot(one,two)
```

```python
histo_x, histo_y = ptu_meas.calculate_g2(correlation_window=50000*1e-12, resolution=600*1e-12, mode='symmetric',channel_start=1,channel_stop=2)
plt.plot(histo_x,histo_y)
```
